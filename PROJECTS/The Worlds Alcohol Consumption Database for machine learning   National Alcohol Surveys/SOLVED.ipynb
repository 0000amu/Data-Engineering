```python
import pandas as pd
import numpy as np 
```

```python
df=pd.read_csv("E:/py/datasets1/project/world_alcohol_comsumption.csv")
df
df
```

```python
#Write a Python program to find out the alcohol consumption details in the year '1987' or 
#'1989' from the dataset. 

res=df[(df['Year']==1987) | (df['Year']==1989)]
res
len(res)
```

```python
#Write a python program to find out and display horizontal bar chart the alcohol 
#consumption details in the year '1987' where WHO region is 'Western Pacific' and country 
#is 'VietNam' from thedataset

res = df[df['Year']==1987].where((df['WHO region']=='Western Pacific')&(df['Country']=='Viet Nam'))
res = res.dropna()
res

```

```python
#horizontal bar chart 
import matplotlib.pyplot as plt 
plt.barh("Beverage Types","Display Value",data=res, color='orange',label='Year 1987')  #bar means vertical barh means horizontal bar 
plt.ylabel("VALUES")
plt.show()
```

```python
#Write a Pandas program to find and display bar chart  to out the alcohol consumption 
#details in the year '1986' or '1989' where WHO region is 'Americas' or 'Europe' from the 
#world alcohol consumption dataset. Use Different colors for each bar and attach text label 
#above each bar. 

res = df[(df.Year==1986) | (df.Year==1989)].where((df['WHO region']=='Americas')
                                            | (df['WHO region']=='Europe'))
res = res.dropna()
res
```

```python
#Write a Pandas program to find out the records where consumption of beverages per 
#person average >=5 and Beverage Types is Beer from world alcohol consumption dataset.

df[(df['Display Value']>=5) & (df['Beverage Types']=='Beer')]
```

```python
#Write a Pandas program to find out and display Pie chart of the records where 
#consumption of beverages per person average >=4 and Beverage Types is Beer, Wine, 
#Spirits from world alcohol consumption dataset.

#res1= df[(df['Display Value']>=4) & (df['Beverage Types'].isin(['Beer','Wine','Spirits']))]
r1 = df[(df['Display Value'] >= 4) & 
        (df['Beverage Types'] == 'Beer')]['Display Value']

r2 = df[(df['Display Value'] >= 4) & 
        (df['Beverage Types'] == 'Wine')]['Display Value']

r3 = df[(df['Display Value'] >= 4) & 
        (df['Beverage Types'] == 'Spirits')]['Display Value']                             
```

```python
plt.pie(r1,shadow=True,autopct='%.1f%%')
plt.show()
plt.pie(r2,shadow=True,autopct='%.1f%%')
plt.show()
plt.pie(r3,shadow=True,autopct='%.1f%%')
plt.show()
```
