```python
#instead of inferSchema use StructType StructField

from pyspark.sql.types import StructField , StructType , StringType , IntegerType
```

```python
from pyspark.sql.types import StructType, StructField, StringType, IntegerType

demo_schema = StructType([
    StructField("emp_id", StringType(), True),
    StructField("name", StringType(), True),
    StructField("department", StringType(), True),
    StructField("salary", IntegerType(), True),
    StructField("city", StringType(), True)
])


df = spark.read \
    .format("csv") \
    .option("header", True) \
    .schema(demo_schema) \
    .load("/Volumes/workspace/default/volume/employees (1).csv")

display(df)  
```

```python
df.write.format('csv') \
    .save('/Volumes/workspace/default/output/employees.csv')
```

```python
#convert csv to json 
df.write.format('json') \
    .option('header',True) \
        .save('/Volumes/workspace/default/output/json_employees')
```
