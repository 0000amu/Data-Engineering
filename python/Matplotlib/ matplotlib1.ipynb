```python
#pie chart 
x = df['marks']
```

```python
s1 =len(x[x>950])
s2 = len(x[(x>850)  &  (x<=950)])
s3 = len(x[x<=850])
print(s1,s2,s3)
```

```python
slices = [s1,s2,s3]
cols = ["pink","red","black"]
lbls = ["above 950","850 to 950","less than 850"]
plt.pie(slices,colors = cols , labels = lbls)
plt.show()
```

```python
#start angel in pie chart using deggrees
slices = [s1,s2,s3]
cols = ["pink","red","maroon"]
lbls = ["above 950","850 to 950","less than 850"]
plt.pie(slices,colors = cols , labels = lbls, startangle=90 , shadow = True , 
       autopct = '%.1f%%' ,explode = (0.5,0,0)
       )
plt.show()
```

```python
# sub plot accomodate many plot side by side 
# first dive page in cells 
x = df['marks']
y = df['rank']
z = df['admit']


```

```python
fig = plt.figure(figsize =( 10,10))
g1 = fig.add_subplot(2,2,1, facecolor ='black')
g1.plot(x,y)
g1.set_title('marks vs rank')

g2 =fig.add_subplot(2,2,2, facecolor='black')
g2.plot(x,z)
g2.set_title('marks vs admit')

g3 = fig.add_subplot(2,2,3, facecolor='black')
high = max(x)  #maxi marks
low = min(x)   #mini marks 
n = len(z[z==1])  # no of students admitted 
g3.pie([high,low,n],colors=['red','pink','black'], labels = lbls,autopct = '%.1f%%' ,explode = (0,0,0))
g3.set_title("marks vs rank vs admission")
plt.show()
```
