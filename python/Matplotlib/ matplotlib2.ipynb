```python
import matplotlib.pyplot as plt
import numpy as np

emp_ages = [
    22, 24, 25, 25, 26, 27, 28, 28, 29, 30,
    30, 31, 32, 32, 33, 34, 35, 35, 36, 38,
    40, 41, 42, 45, 46, 48, 50, 52
]

bins = np.arange(20, 56, 5)

plt.figure(figsize=(10, 6))

plt.hist(
    emp_ages,
    bins=bins,
    histtype='bar',       # bar / barstacked / step / stepfilled
    rwidth=0.85,
    alpha=0.85,
    edgecolor='black',
    linewidth=1.2,
    label='Employee Ages'
)

plt.xlabel("Employee Age", fontsize=12, fontweight='bold')
plt.ylabel("Number of Employees", fontsize=12, fontweight='bold')

plt.title(
    "Employee Age Distribution - Pune AVD",
    fontsize=16,
    fontweight='bold',
    pad=15
)

plt.xticks(bins)

plt.grid(
    axis='y',
    linestyle='--',
    alpha=0.4
)

plt.legend(loc='upper right')

plt.tight_layout()

plt.show()
```
