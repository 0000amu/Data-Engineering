```python
def calculate_salary(basic_sala):
    HRA = basic_sala*20/100
    DA = basic_sala*10/100
    bomus = 5000
    total = bomus+DA+HRA
    print("Total salary =", total)

calculate_salary(50000)
```

```python
def calculate_bill(amount, tax=18):
    final_bill_amount = amount + (amount*tax/100)
    return final_bill_amount

calculate_bill(5000)
```

```python
def calculate_total(price ,quantity ,discount):
    sub_total = price*quantity      #5000
    discount = sub_total * discount/100 #5000 - 10%
    total = sub_total - discount

    print("Total =", total)

calculate_total(1000, 5, 10)
```

```python
def calculate_grade(marks):
    if 90 <= marks <= 100:
        return "A"
    elif 75 <= marks < 90:
        return "B"
    elif 60 <= marks < 75:
        return "C"
    elif 40 <= marks < 60:
        return "D"
    elif 0 <= marks < 40:
        return "F"
    else:
        return "Invalid marks"

print(calculate_grade(100))
```
