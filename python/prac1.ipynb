```python
idly = 60
dosa = 100
vada = 80
puri = 120
cgst = 2.5/100
n = int(input('enter choice: '))
if n <= 4 :
    if n == 1:
        price = idly
        print('idly cost is :',price)
    elif n ==2:

        plain_dosa = 100
        onion_dosa = 120
        masala_dosa = 140

        print("1. Plain Dosa")
        print("2. Onion Dosa")
        print("3. Masala Dosa")

        dosa_choice = int(input("1. Plain Dosa,2. Onion Dosa,3. Masala Dosa,dosa type : "))

        if dosa_choice == 1:
            price = plain_dosa
            print('plain dosa is :',price)

        elif dosa_choice == 2:
            price = onion_dosa
            print('onion dosa is :',price)

        elif dosa_choice == 3:
            price = masala_dosa
            print('masala_dosa cost is :',price)

        else:
            print("Invalid dosa choice")
            exit()

    elif n ==3:
        price = vada
        print('vada cost is :',price)

    else:
        price = puri
        print('puri cost is :',price)
else :
    print('invalid input')

quantity = int(input('enter quantity:'))
total = quantity * price
print('total cost is :',total)

tax  = cgst * total * 2
print('tax cost is :',tax)
grand_total = total + tax
print('grand total is :',grand_total)

paid_amount = int(input('enter paid amount:'))
print('paid amount is :',paid_amount)
balance = paid_amount - grand_total
print('balance is :',balance)
```
