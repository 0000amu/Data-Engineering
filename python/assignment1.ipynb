```python
''' Consider Pizza cost is 326.50, customer order for 3 pizzas, now calculate
bill amount. Shop provides 20% discount on bill amount. So find discount
amount and final Bill '''
pizza_cost = 326.50
order_qyt = 3
bill_amount = order_qyt * pizza_cost
print(bill_amount)
discount_amount = bill_amount / 100 * 20
print(discount_amount)
final_bill = bill_amount - discount_amount
print("Final Bill:", final_bill)
```

```python
'''Read Employee annual salary and hike percentage, then calculate annual
salary after hike. And display new monthly salary '''

salary = int(input("Enter your employee annual salary: "))
hike = int(input("Enter your hike percentage: "))
annual_sala = salary * hike / 100
monthly_salary = annual_salary / 12

print(annual_sala)
print(monthly_salary)
```

```python
'''7. Read Mongo cost per kg, then read quantity. Big Basket gives offer as 22%  discount on Bill Amount.
 So find Total Bill Amount  '''
mango = int(input("Enter your mongo cost: "))
quantity = int(input("Enter your mongo quantity: "))
bill_amount = mango * quantity
big_basket_discount = bill_amount * 22 / 100
final_bill = bill_amount - big_basket_discount
print("Final Bill:", final_bill)
print(big_basket_discount)
print(bill_amount)
```
