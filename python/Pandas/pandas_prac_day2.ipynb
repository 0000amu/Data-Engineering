```python
import pandas as pd 
```

```python
df = pd.read_csv("E:/py/datasets1/empdata.csv")
df
```

```python
df['new_sal']=df['sal'].map(lambda x: x+(x*15/100))
df['new_sal']
df
```

```python
def salary_category(x):
    if x>25000:
        return 'high'
    elif x>15000:
        return 'medium'
    else:
        return 'low'

df['sal_cat']=df['sal'].apply(salary_category)
df
```

```python
def uppercase(x):
    return x.upper()
df['name_mapper']=df['ename'].map(uppercase)
df
```

```python
avg_sala= df['sal'].mean()

def avg_sal(x):
    if x>avg_sala:
        return 'Above avg'
    else:
        return'below avg'
    
df['salary_status']=df['sal'].apply(avg_sal)
df
```

```python
data = {
    "emp": ["A", "A", "B", "B", "C", "C"],
    "month": ["Jan", "Feb", "Jan", "Feb", "Jan", "Feb"],
    "sales": [100, 200, 300, 400, 500, 600]
}
df = pd.DataFrame(data)
df
```

```python
result =df.pivot(
    index='emp',
    columns='month',
    values='sales'
)
print(result)
```

```python
data = {
    "department": ["DE", "DE", "DE", "QA", "QA", "HR"],
    "employee": ["A", "B", "C", "D", "E", "F"],
    "salary": [50000, 60000, 70000, 40000, 50000, 45000]
}
df=pd.DataFrame(data)
df
```

```python
result= df.pivot_table(
    index='department', #rows
    values='salary', #column to calculate 
    aggfunc='mean'  #avg
)

print(result)
```

```python
result= df.pivot_table(
    index='department', #rows
    values='salary', #column to calculate 
    aggfunc=['max','min','mean']  #avg
)

print(result)
```

```python
dataM = {
    "empid": [101, 102, 103],
    "Jan": [100, 200, 300],
    "Feb": [150, 250, 350],
    "Mar": [200, 300, 400]
}
df = pd.DataFrame(dataM)
df
```

```python
result =df.melt(
    id_vars='empid',  #column you keep fixed
    var_name='month',  #name for the old column names
    value_name='sales' #name for the values
    
)
print(result)
```

```python
df=pd.DataFrame(result)
df
```
