```python
import pandas as pd

df = pd.DataFrame({
    "date": pd.date_range("2025-01-01", periods=10),
    "sales": [100, 150, 120, 180, 200, 250, 220, 300, 280, 350]
})

df
```

```python
df['prv_sales']=df['sales'].shift(1)
df
```

```python
df['sales_difference']= df['sales']-df['prv_sales']
```

```python
df['moving_avg']=df['sales'].rolling(3).mean()
```

```python
df['moving_max']=df['sales'].rolling(3).max()
```

```python
df['moving_min']=df['sales'].rolling(3).min()
```

```python
df['running_total']=df['sales'].expanding().sum()
```

```python
df['running_avg']=df['sales'].expanding().mean()
```

# EXPONENTIALLY WEIGHTED MEAN() ewm()

```python
df['ewm_avg']=df['sales'].ewm(span=3).mean()
```

```python
df['growth_perc']= df['sales']-df['prv_sales']/df['prv_sales']* 100
```

# timefunction 

```python
import pandas as pd

df = pd.DataFrame({
    "date": pd.date_range("2025-01-01", periods=90, freq="D"),
    "sales": range(100, 190)
})

df
```

```python
df['date']=pd.to_datetime(df['date'])
```

```python
df['date']
```

```python
df['year']=df['date'].dt.year
df['month']=df['date'].dt.month
df['day']=df['date'].dt.day
df['day_name']=df['date'].dt.day_name()
df['quarter']=df['date'].dt.quarter
```

```python
df
```

```python
df[df['date']>='2025-02-01']
```

```python
df
```

```python
df = df.set_index('date')

monthly_sales = df['sales'].resample('ME').sum()

print(monthly_sales)
```

```python
monthly_sales_avg = df['sales'].resample('ME').mean()

print(monthly_sales_avg)
```

```python
Calculate_monthly =df['sales'].resample('ME').agg(['sum','mean','max','min'])
print(Calculate_monthly)
```

```python
quarter_sales= df['sales'].resample('QE').agg(sum)
print(quarter_sales)
```

```python
weekly_sales = df['sales'].resample('W').sum()
print(weekly_sales)
```
