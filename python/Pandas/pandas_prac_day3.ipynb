```python
import pandas as pd
```

```python

employees = pd.DataFrame({
    "empid": [101, 102, 103, 104],
    "ename": ["Amey", "Raj", "Amit", "Suresh"]
})

df1 = pd.DataFrame(employees)
df1
```

```python

salary = pd.DataFrame({
    "empid": [101, 102, 103, 105],
    "sal": [50000, 60000, 70000, 80000]
})
df2=pd.DataFrame(salary)
df2
```

```python
df3=pd.merge(employees,salary, on='empid',how='inner')
df3
```

```python
df4=pd.merge(employees,salary,on='empid',how='left')
df4
```

```python
df5=pd.merge(employees,salary,on='empid',how='outer',indicator=True)
df5
```

```python
df2.rename(columns={'empid':'employee_id'},inplace=True)
df2
```

```python
df6=pd.merge(df1,df2,left_on='empid',right_on='employee_id')
df6
```

```python
df = pd.DataFrame({
    "month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
    "sales": [100, 200, 150, 300, 250, 400]
})
df
```

```python
df['previous_month_sales']=df['sales'].shift(1)
df['previous_month_sales']
df
```

```python
df['sales_change']= df['sales']-df['previous_month_sales']
df
```

```python
df['3_month_avg']= df['sales'].rolling(3).mean()
```

```python
df
```

```python
df['running_total']=df['sales'].cumsum()
```

```python
df['running_avg']=df['sales'].expanding().mean()
```
