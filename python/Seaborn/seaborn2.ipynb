```python
import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sns  
```

```python
df = sns.load_dataset('iris')
```

```python
#pair plot represent replationship between all columns 
sns.pairplot(data = df ,hue = 'species' , palette = 'Blues' , markers=['*','X','s']) 
# markers=>> o, d , s , X ,* , v 
#palette : BrBG, RdBu, Blues ,husl , coolwarm , cubehelix 
```

```python
#relation between only 2 columns 
sns.pairplot(df, vars=['sepal_length','sepal_width'])
```

### heat map 
df1 = df.iloc[:10,:4]
sns.heatmap(df1,annot=True, cmap='Blues')

```python
#cat plot 
#kind = violin , boxen
sns.catplot(x='species',y='sepal_length',data=df,cmap='Blues',kind ='boxen')
```
