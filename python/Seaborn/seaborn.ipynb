```python
import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sns  
```

```python
df = sns.load_dataset('mpg')
```

```python
df.head()
```

```python
df['cylinders'].unique()
```

```python
# distribution plot    and tale it as colm covert into typle and then give it 
tpl = tuple(df['cylinders'])
sns.displot(tpl, kde=True, bins=range(3,10,1),color = )
plt.xlabel('no of cylinders')
plt.xlabel('no of cars')
plt.show()
```

```python
#count plot
sns.countplot(x='cylinders', data = df, palette = 'dark:red', hue = 'origin')
plt.legend(loc= 'lower right')
```

```python
sns.boxplot(x='origin', y ='acceleration',data = df, palette = 'dark:red')
```

```python
#joined plot
sns.jointplot(x='horsepower', y='acceleration',data = df, palette = 'dark:red' , hue='origin')
```

```python
#scatter plot 
sns.scatterplot(x='horsepower',y='acceleration',data = df ,alpha= 0.5 )
```

```python
#sub plot more than one plot in same pqg e
```

```python
plt.figure(figsize=(10,10))
plt.subplot(2,2,1)
sns.scatterplot(x='horsepower',y='acceleration',data = df , hue = 'origin',palette = 'dark:red')

plt.subplot(2,2,2)
sns.scatterplot(x='origin',y='acceleration',data=df ,palette = 'dark:red')

plt.subplot(2,2,3)
sns.boxplot(x='origin', y ='acceleration',data = df, palette = 'dark:red')


plt.savefig('E:/py/datasets1/myfig.png')
```
